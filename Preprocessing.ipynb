```python
import pandas as pd
import seaborn as sns
from google.colab import files
import numpy as np
```

```python
# Mengimpor dataset
df = pd.read_csv('/content/midterm_hotel_data.csv')
df
```

```python
# Menampilkan informasi awal untuk melihat tipe data
print("\nInformasi Tipe Data Awal:")
print(df.info())

# Mengecek data duplikat
duplicate_count = df.duplicated().sum()
print(f"\nJumlah data duplikat: {duplicate_count}")

# Mengecek data kosong
missing_values = df.isnull().sum()
print("\nJumlah data kosong di setiap kolom:")
print(missing_values)
```

```python
# Menghitung statistik deskriptif untuk kolom numerik
desc_stats = df.select_dtypes(include=['float64', 'int64']).agg(['mean', 'median', 'std', 'min', 'max', 'count']).T

# Menambahkan mode secara terpisah (karena mode bisa memiliki lebih dari satu nilai)
desc_stats['mode'] = df.select_dtypes(include=['float64', 'int64']).mode().iloc[0]

# Menampilkan hasil
desc_stats
```

```python
# Mengisi missing value pada kolom yang memiliki kosong dengan mean/median
# Kolom lead_time diisi dengan median
df.loc[:, 'lead_time'] = df['lead_time'].fillna(df['lead_time'].median())
# Kolom stays_in_weekend_nights diisi dengan median
df.loc[:, 'stays_in_weekend_nights'] = df['stays_in_weekend_nights'].fillna(df['stays_in_weekend_nights'].median())
# Kolom adults diisi dengan median
df.loc[:, 'adults'] = df['adults'].fillna(df['adults'].median())
# Kolom country diisi dengan modus
df.loc[:, 'country'] = df['country'].fillna(df['country'].mode()[0])
# Kolom children diisi dengan median
df.loc[:, 'children'] = df['children'].fillna(df['children'].median())
# Kolom adr diisi dengan median
df.loc[:, 'adr'] = df['adr'].fillna(df['adr'].median())
# Kolom total_of_special_requests diisi dengan median
df.loc[:, 'total_of_special_requests'] = df['total_of_special_requests'].fillna(df['total_of_special_requests'].median())
# Kolom company diisi dengan median
df.loc[:, 'company'] = df['company'].fillna(df['company'].median())
# Kolom agent diisi dengan 0
df.loc[:, 'agent'] = df['agent'].fillna(0)

# Menampilkan DataFrame yang diperbarui
print("\nDataFrame setelah mengisi nilai kosong pada kolom yang memiliki nilai kosong:")
print(df[['lead_time', 'stays_in_weekend_nights', 'adults', 'country', 'children', 'adr', 'total_of_special_requests', 'company', 'agent']])

# Mengecek kembali jumlah data kosong setelah pengisian
missing_values_after = df.isnull().sum()
print("\nJumlah data kosong di setiap kolom setelah pengisian:")
print(missing_values_after)
```

```python
# MENGECEK OUTLIER SETELAH PENGISIAN MISSING VALEUE
def detect_outliers_iqr(data):

  Q1 = data.quantile(0.25)
  Q3 = data.quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = data[(data < lower_bound) | (data > upper_bound)]
  return outliers

for column in df.columns:
  if pd.api.types.is_numeric_dtype(df[column]):
    outliers = detect_outliers_iqr(df[column])
    if not outliers.empty:
      print(f"Outliers found in column '{column}':")
      print(outliers)
    else:
      print(f"No outliers found in column '{column}'")
```

```python
# Menghapus kolom yang tidak mempunyai nama kolom
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
df
```

```python
# Menyimpan DataFrame yang sudah diolah ke dalam file CSV
output_file_path = '/content/processed_hotel_data.csv'
df.to_csv(output_file_path, index=False)
# Mengunduh file CSV yang sudah disimpan
files.download(output_file_path)
```
