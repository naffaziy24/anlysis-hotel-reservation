```python
df = pd.read_csv("/content/preprocessing_hotel_data.csv")
df
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Menghitung total lead time dan total pembatalan
total_lead_time = df['lead_time'].sum()
total_cancellations = df['is_canceled'].sum()

# Menghitung total keseluruhan
total = total_lead_time + total_cancellations

# Menghitung persentase untuk masing-masing
lead_time_percentage = (total_lead_time / total) * 100
cancellations_percentage = (total_cancellations / total) * 100

# Menyusun data untuk pie chart
labels = ['Total Lead Time', 'Jumlah Reservasi Dibatalkan']
sizes = [lead_time_percentage, cancellations_percentage]

# Membuat pie chart
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title("Rasio Total Lead Time dengan Jumlah Reservasi yang Dibatalkan")
plt.show()
```

# Modul 4 no 2

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Konversi kolom `arrival_date_month` menjadi kategori agar urutannya benar
data['arrival_date_month'] = pd.Categorical(
    data['arrival_date_month'],
    categories=['January', 'February', 'March', 'April', 'May', 'June', 'July',
                'August', 'September', 'October', 'November', 'December'],
    ordered=True
)

# Menambahkan kolom pendapatan total (adr * (stays_in_week_nights + stays_in_weekend_nights))
data['total_revenue'] = data['adr'] * (data['stays_in_week_nights'] + data['stays_in_weekend_nights'])

# Memisahkan data berdasarkan jenis hotel
resort_data = data[data['hotel'] == 'Resort Hotel']
city_data = data[data['hotel'] == 'City Hotel']

# Menghitung pendapatan bulanan untuk setiap jenis hotel
resort_revenue = resort_data.groupby('arrival_date_month')['total_revenue'].sum()
city_revenue = city_data.groupby('arrival_date_month')['total_revenue'].sum()

# Membuat DataFrame untuk hasil analisis
revenue_comparison = pd.DataFrame({
    'Pendapatan Resort Hotel': resort_revenue,
    'Pendapatan City Hotel': city_revenue
})

# Tampilkan tabel perbandingan pendapatan
print("Perbandingan Pendapatan Bulanan Antara Resort Hotel dan City Hotel:")
print(revenue_comparison)

# Visualisasi perbandingan pendapatan bulanan
plt.figure(figsize=(14, 7))
plt.plot(revenue_comparison.index, revenue_comparison['Pendapatan Resort Hotel'], marker='o', label='Resort Hotel', color='b')
plt.plot(revenue_comparison.index, revenue_comparison['Pendapatan City Hotel'], marker='o', label='City Hotel', color='g')
plt.title('Perbandingan Pendapatan Bulanan Antara Resort Hotel dan City Hotel')
plt.xlabel('Bulan')
plt.ylabel('Pendapatan')
plt.legend()
plt.grid(True)
plt.show()
```

# Modul 4 no 3

```python
import pandas as pd
import matplotlib.pyplot as plt

# Filter data untuk reservasi yang dibatalkan
canceled_data = data[data['is_canceled'] == 1]

# Mengelompokkan berdasarkan negara dan jenis hotel untuk menghitung jumlah pembatalan
cancellation_by_country = canceled_data.groupby(['country', 'hotel']).size().reset_index(name='cancellation_count')

# Mengambil 10 negara teratas berdasarkan jumlah pembatalan
top_countries = cancellation_by_country.groupby('country')['cancellation_count'].sum().nlargest(10).index
top_cancellation_data = cancellation_by_country[cancellation_by_country['country'].isin(top_countries)]

# Membuat bar chart
plt.figure(figsize=(12, 8))
for hotel_type in top_cancellation_data['hotel'].unique():
    hotel_data = top_cancellation_data[top_cancellation_data['hotel'] == hotel_type]
    plt.bar(hotel_data['country'], hotel_data['cancellation_count'], label=hotel_type)

# Menambahkan detail pada grafik
plt.title("10 Negara Teratas dengan Pembatalan Reservasi Tertinggi untuk Resort Hotel dan City Hotel")
plt.xlabel("Negara")
plt.ylabel("Jumlah Pembatalan Reservasi")
plt.legend(title="Jenis Hotel")
plt.xticks(rotation=45)
plt.show()
```

# Modul 4 no 4

```python
# Filter untuk pemesanan last minute (lead_time <= 7)
last_minute_bookings = data[data['lead_time'] <= 7]

# 1. Menghitung tingkat pembatalan untuk pemesanan last minute
cancellation_rate = last_minute_bookings['is_canceled'].mean() * 100
print(f"Tingkat pembatalan untuk pemesanan last minute: {cancellation_rate:.2f}%")

# 2. Analisis distribusi lead time
lead_time_distribution = last_minute_bookings['lead_time'].value_counts().sort_index()
print("Distribusi lead time untuk pemesanan last minute:")
print(lead_time_distribution)

# 3. Analisis berdasarkan jenis pelanggan
customer_type_analysis = last_minute_bookings['customer_type'].value_counts()
print("Analisis jenis pelanggan untuk pemesanan last minute:")
print(customer_type_analysis)

# 4. Analisis berdasarkan saluran distribusi
distribution_channel_analysis = last_minute_bookings['distribution_channel'].value_counts()
print("Analisis saluran distribusi untuk pemesanan last minute:")
print(distribution_channel_analysis)

# 5. Visualisasi
# Visualisasi tingkat pembatalan
plt.figure(figsize=(10, 5))
plt.bar(['Pembatalan', 'Tidak Pembatalan'], [last_minute_bookings['is_canceled'].sum(), len(last_minute_bookings) - last_minute_bookings['is_canceled'].sum()], color=['red', 'green'])
plt.title('Tingkat Pembatalan Pemesanan Last Minute')
plt.ylabel('Jumlah Pemesanan')
plt.xticks(rotation=0)
plt.show()

# Visualisasi distribusi lead time
plt.figure(figsize=(10, 6))
lead_time_distribution.plot(kind='bar', color='skyblue')
plt.title('Distribusi Lead Time untuk Pemesanan Last Minute')
plt.xlabel('Lead Time (Hari)')
plt.ylabel('Jumlah Pemesanan')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()

# Visualisasi jenis pelanggan
plt.figure(figsize=(10, 5))
customer_type_analysis.plot(kind='bar', color='orange')
plt.title('Analisis Jenis Pelanggan untuk Pemesanan Last Minute')
plt.xlabel('Jenis Pelanggan')
plt.ylabel('Jumlah Pemesanan')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()

# Visualisasi saluran distribusi
plt.figure(figsize=(10, 5))
distribution_channel_analysis.plot(kind='bar', color='purple')
plt.title('Analisis Saluran Distribusi untuk Pemesanan Last Minute')
plt.xlabel('Saluran Distribusi')
plt.ylabel('Jumlah Pemesanan')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```
